# Loops in Python

- For Repeating tasks

## Types of Loops :

- For loop
- While Loop

```python
# If we print normal statement 10 times 
print("Usman Ghani")
print("Usman Ghani")
print("Usman Ghani")
print("Usman Ghani")
print("Usman Ghani")
print("Usman Ghani")
print("Usman Ghani")
print("Usman Ghani")
print("Usman Ghani")
print("Usman Ghani")
# It is very complex and line wasting thing.
```

```python
# So we use Loops
# - For Loop
for i in range(10):
    print(f"{i+1}: Usman Ghani")
```

```python
# Type Casting
list(range(10))
```

```python
# List
students = ["Usman" , "Hammad" , "Ahsan" , "Javed" , "Ubaid"]
for i in range(5):
    print(f"Hello! {students[i]}")
```

```python
print(students)
```

```python
# Iterating over list
for s in students:
    print(f"Hello! {s}")
```

# Break

```python
print(students)
```

```python
for student in students:
    if student == "Javed":
        break
    print(f"Hello! {student}")
```

# Continue

```python
for u in range(10):
    if u == 5:
        continue
    print(f"{u} : Hello World")
```
